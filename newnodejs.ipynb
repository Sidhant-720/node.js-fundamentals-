Q4:- Write a script that accesses and displays the command-line arguments passed to it when executed.
* Ans:-

````python
// args.js
const args = process.argv.slice(2);

console.log("Arguments:", args);

// in terminal
// node args.js hello world 42
````

 `process.argv` is an array where index 0 is the path to Node, index 1 is the path to your script, and everything after that is your arguments. `slice(2)` drops the first two. All values come in as strings, so convert with `Number(args[2])` if you need a number.
To wrap an argument containing spaces, use quotes: `node args.js "hello world"`.

Q5:- Using the fs/promises module, write a script to asynchronously read the content of a file named data.txt and log it to the console.
* Ans:-




````python
// read.js
const fs = require("fs/promises");

async function readFile() {
  try {
    const content = await fs.readFile("data.txt", "utf8");
    console.log(content);
  } catch (err) {
    console.error("Error reading file:", err.message);
  }
}

readFile();
````

* in terminal:-
echo "Some sample text" > data.txt node read.js
* Passing `"utf8"` returns a string; without it you get a raw `Buffer`. The path is relative to where you run the command, not where the script lives. If the file is missing, the `catch` block logs something like `ENOENT: no such file or directory`.

 Question 6: Using the fs/promises module, write a script to create a new file named log.txt and write the string "Task completed." into it.
 * Ans:-





````python
// write.js
const fs = require("fs/promises");

async function writeLog() {
  try {
    await fs.writeFile("log.txt", "Task completed.");
    console.log("log.txt created.");
  } catch (err) {
    console.error("Error writing file:", err.message);
  }
}

writeLog();
````

Run it in terminal:-
node write.js
cat log.txt

`writeFile` creates the file if it doesn't exist and overwrites it if it does. To add to the end instead, use `fs.appendFile("log.txt", "Task completed.\n")`.

Q7:- Use the path module to safely join the segments src, utils, and helpers.js into a single, cross-platform-compatible file path.

Q8:- Create a basic HTTP server using the http module that listens on port 3000 and responds to all requests with "Server is running."
* Ans:-


````python
// server.js
const http = require("http");

const server = http.createServer((req, res) => {
  res.writeHead(200, { "Content-Type": "text/plain" });
  res.end("Server is running.");
});

server.listen(3000, () => {
  console.log("Listening on http://localhost:3000");
});
````

Run it:
node server.js
* Test it (in a second terminal, or just open the URL in a browser):
`curl http://localhost:3000`
* The callback runs for every incoming request, regardless of path or method. Stop the server with Ctrl+C. If you see EADDRINUSE, something else is already using port 3000.

Q9:- Enhance the previous HTTP server to handle two different routes: / should respond with "Homepage", and /about should respond with "About Us".
* Ans:-


req.url holds the path of the request, so a simple if/else chain routes it. The final else returns a 404 for anything unmatched. Note that req.url includes query strings (/about?x=1 wouldn't match), so for real apps you'd parse it with new URL(req.url, "http://localhost") or use a framework like Express.

````python
// server.js
const http = require("http");

const server = http.createServer((req, res) => {
  if (req.url === "/") {
    res.writeHead(200, { "Content-Type": "text/plain" });
    res.end("Homepage");
  } else if (req.url === "/about") {
    res.writeHead(200, { "Content-Type": "text/plain" });
    res.end("About Us");
  } else {
    res.writeHead(404, { "Content-Type": "text/plain" });
    res.end("Not Found");
  }
});

server.listen(3000, () => {
  console.log("Listening on http://localhost:3000");
});
````

* Run it:
node server.js
* Test it:
curl http://localhost:3000/
curl http://localhost:3000/about
curl http://localhost:3000/contact
* `req.url` holds the path of the request, so a simple `if/else` chain routes it. The final `else` returns a 404 for anything unmatched. Note that `req.url` includes query strings (`/about?x=1` wouldn't match), so for real apps you'd parse it with `new URL(req.url, "http://localhost")` or use a framework like Express.

Q10:-Write a script that efficiently reads a large file (large.txt) and outputs its content to the console using streams.
* Ans:-

````python
const fs = require("fs");

const stream = fs.createReadStream("large.txt", { encoding: "utf8" });

stream.pipe(process.stdout);

stream.on("error", (err) => {
  console.error("Error reading file:", err.message);
});
````

Run it:
node stream-read.js

A stream reads the file in small chunks (64 KB by default) instead of loading the whole thing into memory like `readFile` does, so memory use stays flat even for multi-gigabyte files. `pipe` sends each chunk to `process.stdout` and handles backpressure for you, pausing the read if the console can't keep up.
To process chunks yourself instead of piping:


````python
//js
stream.on("data", (chunk) => console.log(chunk));
stream.on("end", () => console.log("Done."));
````
